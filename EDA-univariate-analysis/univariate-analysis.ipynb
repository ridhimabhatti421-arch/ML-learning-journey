```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('train.csv')
df.head()
```

## 1. CATEGORICAL DATA

### a. Countplot

```python
sns.countplot(df['Embarked'])
```

```python
sns.countplot(data=df,x='Survived' )
```

### b. PieChart

```python
df['Sex'].value_counts().plot(kind='pie',autopct='%.2f')
```

## 2. NUMERICAL DATA

### a. Histogram

```python
plt.hist(df['Age'],bins=5)
```

### b. Displot

```python
sns.displot(df['Age'], kind='kde')
```

### c. Boxplot

```python
sns.boxplot(df['Age'])
```

```python
# to calculate  nim value
df['Age'].min()
```

```python
# to calculate max age
df['Age'].max()
```
