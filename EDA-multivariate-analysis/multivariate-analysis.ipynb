```python
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
titanic = pd.read_csv('train.csv')
titanic
```

```python
flights = sns.load_dataset('flights')
flights
```

```python
iris = sns.load_dataset('iris')
iris
```

## 1. Scatterplot (Numerical - Numerical)

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='smoker')
```

## 2. Bar Plot(numerical -categorical)

```python
sns.barplot(data=titanic,y='Age', x='Pclass',hue='Sex')
```

## 3. Box Plot (Numerical - Categorical)

```python
sns.boxplot(data=titanic,x='Sex', y='Age',hue='Survived')
```

## 4. HeatMap (Categorical - Categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

## 5. ClusterMap (Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

## 6. Pairplot

```python
sns.pairplot(iris,hue='species')
```

## 7. Lineplot (Numerical - Numerical)

```python
sns.lineplot(data=flights, x='year', y='passengers')
```

```python
sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'))
```
